```python
import pandas as pd

single_result_path = "../results/tplm_benchmark_results"
fusion_result_path = "../results/embedding_fusion_results"

results = pd.DataFrame()
```

```python
for dataset in ['aav','gb1','gfp','location','meltome','stability']:
    single_df =  pd.read_csv(f"{single_result_path}/{dataset}_log.tsv", sep='\t')
    fusion_df = pd.read_csv(f"{fusion_result_path}/{dataset}_log.tsv", sep='\t')

    results = pd.concat([results, single_df, fusion_df])
    
```
